### 2.4 Outlier Check

```python
cols = ['GA0151_A', 'GA0151_C', 'GA0151_D']
Q1 = df[cols].quantile(0.25)
Q3 = df[cols].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

outlier = (df[cols] < lower) | (df[cols] > upper)
print("Outlier per sensor:\n", outlier.sum())
```
